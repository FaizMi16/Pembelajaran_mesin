## C. Langkah Latihan
### Langkah 1 — Mengenali fitur & label

```python

import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]} # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']] # fitur
y = df['harga'] # label
```

### Langkah 2 — Klasifikasi vs regresi (amati tipe label).

```python
print(df['harga'].tolist())

df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']
```

### Langkah 3 — Pembagian data latih & uji.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))
```

## D. Latihan Mandiri
### Soal 1 — DataFrame Siswa

```python
import pandas as pd

data_siswa = {
    'jam_belajar': [2, 5, 1, 7, 3, 6],
    'lulus': ['Tidak', 'Lulus', 'Tidak', 'Lulus', 'Tidak', 'Lulus']
}
df_siswa = pd.DataFrame(data_siswa)

X_siswa = df_siswa[['jam_belajar']]
y_siswa = df_siswa['lulus']

print(df_siswa)
```

**Analisis:** Ini termasuk **klasifikasi**, karena label `lulus` berupa kategori
diskrit ("Lulus"/"Tidak"), bukan angka kontinu. Berbeda dengan kasus `harga` pada
contoh dosen yang labelnya numerik (regresi).

### Soal 2 — Split Data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_siswa, y_siswa, test_size=0.2, random_state=42)

print('Latih:', len(X_train), ' Uji:', len(X_test))
```

### Soal 3 — Kasus Prediksi Jumlah Pengunjung Toko

- **Jenis pembelajaran:** Supervised learning (data historis punya jawaban/hasil yang diketahui)
- **Subkategori:** Regresi (jumlah pengunjung = angka kontinu, bukan kategori)
- **Fitur (X):** hari (weekday/weekend), tanggal/bulan, cuaca, ada promo atau tidak, jam operasional
- **Label (y):** jumlah pengunjung besok

## E. Refleksi

**1. Apa perbedaan supervised, unsupervised, dan reinforcement learning?**
- *Supervised*: belajar dari data berlabel untuk memprediksi label baru.
- *Unsupervised*: data tanpa label; model mencari pola/kelompok sendiri.
- *Reinforcement*: belajar lewat coba-coba, dari reward/punishment lingkungan.

**2. Mengapa pembagian data latih dan uji penting?**
Untuk menguji apakah model benar-benar generalisasi ke data baru, bukan cuma menghafal data latih (menghindari overfitting).

**3. Mengapa alur kerja ML disebut iteratif?**
Karena prosesnya bolak-balik: data → latih → evaluasi → perbaiki, diulang sampai model cukup baik.
